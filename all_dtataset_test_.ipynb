```python
import os

datasets_path = "/content/drive/MyDrive/finger vein research dataset/Datasets"

for name in os.listdir(datasets_path):
    path = os.path.join(datasets_path, name)

    if os.path.isdir(path):
        print("\n==============================")
        print("DATASET:", name)
        print("==============================")

        # Top-level files/folders
        items = os.listdir(path)

        print("Number of items:", len(items))
        print("First 15 items:")

        for item in items[:15]:
            print("  ", item)
```

```python
import os

datasets_path = "/content/drive/MyDrive/finger vein research dataset/Datasets"

image_extensions = (".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff")

for dataset_name in os.listdir(datasets_path):

    dataset_path = os.path.join(datasets_path, dataset_name)

    if not os.path.isdir(dataset_path):
        continue

    image_files = []

    for root, dirs, files in os.walk(dataset_path):
        for file in files:
            if file.lower().endswith(image_extensions):
                image_files.append(os.path.join(root, file))

    print("\n==============================")
    print("DATASET:", dataset_name)
    print("==============================")
    print("Total images:", len(image_files))

    if len(image_files) > 0:
        print("\nFirst 5 images:")
        for img in image_files[:5]:
            print(" ", img)
```
